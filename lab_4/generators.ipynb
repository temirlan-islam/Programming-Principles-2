```python
def generator(n):
    i=0
    squares = []
    while i**2<=n**2:
        squares.append(i**2)
        i+=1
    return squares
n=int(input())
print(generator(n))
```

```python
def evens(n):
    result=[]
    for i in range(n):
        if i%2==0:
            result.append(i)
    return result

n=int(input())
k=evens(n)
print(k)
```

```python
def d34(n):
    result=[]
    for i in range(n):
        if i%3==0 and i%4==0:
            result.append(i)
    return result

n=int(input())
k=d34(n)
print(k)
```

```python
def squares(a, b):
    res=[]
    for i in range(a, b + 1):
        res.append(i**2)
    return res
a=int(input())
b=int(input())
t=squares(a,b)
print(t)
```

```python
def countdown(n):
    res=[]
    for i in range(n, -1, -1):
        res.append(i)
    return res

n = int(input())
k=countdown(n)
print(k)
```
