```python
import re

with open('row.txt', 'r', encoding='utf-8') as file:
    content = file.read()

pattern = re.compile(
    r'^\d+\.\n'              
    r'(.+)\n'                
    r'(\d{1,3}(?:,\d{3})*) x (\d+,\d{2})\n'  
    r'(\d+,\d{2})$',  
    re.MULTILINE
)

matches = pattern.findall(content)

products = []
for match in matches:
    name = match[0].strip()
    quantity = float(match[1].replace(',', ''))  
    unit_price = float(match[2].replace(',', '.')) 
    total_price = float(match[3].replace(',', '.'))  
    products.append({
        'name': name,
        'quantity': quantity,
        'unit_price': unit_price,
        'total_price': total_price
    })


print("=" * 30)
print("{:<20}{:>8}".format("Товар", "Сумма"))
print("=" * 30)
for product in products:
    print(f"{product['name']:<20}{product['total_price']:>8.2f} KZT")
print("=" * 30)
print(f"{'Итого':<20}{sum(p['total_price'] for p in products):>8.2f} KZT")
print("=" * 30)
```
